```python
import pandas as pd

df = pd.read_csv("../data/cleaned/diabetic_data_cleaned.csv")
print(df.shape)
print(list(df.columns))
```

```python
for col in df.columns:
    print(col)
```

```python
df = df.sort_values(['patient_nbr', 'encounter_id']).reset_index(drop=True)

df['visit_number'] = df.groupby('patient_nbr').cumcount() + 1
df['total_encounters'] = df.groupby('patient_nbr')['encounter_id'].transform('count')

df[['patient_nbr', 'encounter_id', 'visit_number', 'total_encounters']].head(10)
```

```python
total_patients = df['patient_nbr'].nunique()
repeat_patients = df[df['total_encounters'] > 1]['patient_nbr'].nunique()

print("Total unique patients:", total_patients)
print("Repeat patients (2+ visits):", repeat_patients)
print("Percentage repeat:", round(repeat_patients / total_patients * 100, 2), "%")
```

```python
df['readmitted_flag'] = (df['readmitted'] == '<30').astype(int)

df['readmitted_flag'].value_counts()
```

```python
df.to_csv("../data/cleaned/diabetic_data_features.csv", index=False)
print("Saved! Final shape:", df.shape)
print("New columns added:", ['visit_number', 'total_encounters', 'readmitted_flag'])
```
