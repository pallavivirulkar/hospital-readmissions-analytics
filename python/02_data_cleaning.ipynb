```python
import pandas as pd

df = pd.read_csv("../data/raw/diabetic_data.csv")
df = df.replace('?', pd.NA)

df.isnull().sum().sort_values(ascending=False).head(10)
```

```python
raw = pd.read_csv("../data/raw/diabetic_data.csv")
print(raw['max_glu_serum'].value_counts(dropna=False))
print()
print(raw['A1Cresult'].value_counts(dropna=False))
```

```python
df['max_glu_serum'] = df['max_glu_serum'].fillna('Not Tested')
df['A1Cresult'] = df['A1Cresult'].fillna('Not Tested')

df[['max_glu_serum', 'A1Cresult']].isnull().sum()
```

```python
df = df.drop(columns=['weight'])
df.shape
```

```python
df = df.drop(columns=['payer_code'])
df.shape
```

```python
df['medical_specialty'] = df['medical_specialty'].fillna('Unknown')
df['medical_specialty'].isnull().sum()
```

```python
df['race'] = df['race'].fillna('Unknown')
df['race'].isnull().sum()
```

```python
df.isnull().sum().sort_values(ascending=False).head(10)
```

```python
df = df.dropna(subset=['diag_1'])

df['diag_2'] = df['diag_2'].fillna('None')
df['diag_3'] = df['diag_3'].fillna('None')

print(df.shape)
print(df.isnull().sum().sum())
```

```python
exclude_codes = [11, 13, 14, 19, 20, 21]
before = df.shape[0]

df = df[~df['discharge_disposition_id'].isin(exclude_codes)]

after = df.shape[0]
print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)
```

```python
df['gender'].value_counts()
```

```python
df = df[df['gender'] != 'Unknown/Invalid']
df.shape
```

```python

df.to_csv("../data/cleaned/diabetic_data_cleaned.csv", index=False)
print("Saved! Final shape:", df.shape)
```
